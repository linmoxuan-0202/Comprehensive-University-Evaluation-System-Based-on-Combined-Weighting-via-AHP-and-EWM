# 单一权重

```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
# ============================================================
# 文件路径
# ============================================================
input_file = Path(r"./QS_2026_中国大陆高校官方分项评分.xlsx")
sheet_name = 0
df = pd.read_excel(input_file, sheet_name=sheet_name)


output_file = Path(r"./QS_2026_重新计算总分及排名.xlsx")
```

```python
# ============================================================
# 六项指标权重
# ============================================================
weights = {
    "国际学生比例": 0.05,
    "国际教师比例": 0.05,
    "师生比": 0.20,
    "每位教员引用率": 0.20,
    "学术声誉": 0.40,
    "雇主声誉": 0.10,
}
```

```python


# 重新计算总分
df["新总分"] = sum(
    df[indicator] * weight
    for indicator, weight in weights.items()
)


# 按新总分排名
df["新排名"] = df["新总分"].rank(
    ascending=False,
    method="min"
).astype(int)


# 按新排名排序
df = df.sort_values("新排名")


# 保存结果
df.to_excel(
    "QS_2026_单权重重新排名结果.xlsx",
    index=False
)

print(df[["新排名", "大学", "新总分"]])
```

# 混合权重

```python
# 权重一
weights_1 = {
    "国际学生比例": 0.05,
    "国际教师比例": 0.05,
    "师生比": 0.20,
    "每位教员引用率": 0.20,
    "学术声誉": 0.40,
    "雇主声誉": 0.10
}

# 权重二
weights_2 = {
    "国际学生比例": 0.10,
    "国际教师比例": 0.10,
    "师生比": 0.15,
    "每位教员引用率": 0.25,
    "学术声誉": 0.25,
    "雇主声誉": 0.15
}

# 组合系数
alpha = 0.6

# 计算组合权重
new_weights = {
    indicator: alpha * weights_1[indicator]
               + (1 - alpha) * weights_2[indicator]
    for indicator in weights_1
}

# 输出组合权重
for indicator, weight in new_weights.items():
    print(f"{indicator}: {weight:.4f}")

print("权重之和：", sum(new_weights.values()))
```

```python
df["新总分"] = sum(
    df[indicator] * weight
    for indicator, weight in new_weights.items()
)

df["新排名"] = df["新总分"].rank(
    ascending=False,
    method="min"
).astype(int)

df = df.sort_values("新排名")

df.to_excel("QS_2026_混合权重组合权重排名结果.xlsx", index=False)

print(df[["新排名", "大学", "新总分"]])
```
